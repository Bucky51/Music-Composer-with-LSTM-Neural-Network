```python
from music21 import converter, instrument, note, chord

notes = []

for i, file in enumerate(result):
    print(f'{i+1}: {file}')
    try:
        midi = converter.parse(file)  # Parse the MIDI file
        parts = instrument.partitionByInstrument(midi)  # Check for multiple instrument parts
        if parts:  # File has instrument parts
            notes_to_parse = parts.parts[0].recurse()  # Use the first part
        else:  # File has notes in a flat structure
            notes_to_parse = midi.flat.notes

        # Extract notes and chords from the MIDI file
        for element in notes_to_parse:
            if isinstance(element, note.Note):
                notes.append(str(element.pitch))  # Add pitch of the note
            elif isinstance(element, chord.Chord):
                notes.append('.'.join(str(n) for n in element.normalOrder))  # Add chord as normal order
    except Exception as e:
        print(f'FAILED: {i+1}: {file} due to {e}')  # Print the error if file fails to process

# Check the extracted notes
print(f"Total notes extracted: {len(notes)}")
```

```python
import numpy as np
from tensorflow.keras.utils import to_categorical

def prepare_sequences(notes, n_vocab):
    """Prepare sequences of notes for training."""
    sequence_length = 32  # Length of each sequence

    # Get all unique pitch names (sorted)
    pitchnames = sorted(set(item for item in notes))
    numPitches = len(pitchnames)

    # Create a dictionary to map pitch names to integers
    note_to_int = dict((note, number) for number, note in enumerate(pitchnames))

    network_input = []
    network_output = []

    # Create input sequences and the corresponding outputs
    for i in range(0, len(notes) - sequence_length, 1):
        sequence_in = notes[i:i + sequence_length]  # Input sequence
        sequence_out = notes[i + sequence_length]   # Next note after sequence
        network_input.append([note_to_int[char] for char in sequence_in])  # Convert notes to integers
        network_output.append(note_to_int[sequence_out])  # Corresponding output as integer

    n_patterns = len(network_input)

    # Reshape input into a format compatible with LSTM layers (3D)
    network_input = np.reshape(network_input, (n_patterns, sequence_length, 1))
    # Normalize input
    network_input = network_input / float(n_vocab)

    # One-hot encode the output
    network_output = to_categorical(network_output, num_classes=n_vocab)

    return network_input, network_output

# Ensure the notes list is populated
if len(notes) > 32:
    n_vocab = len(set(notes))  # Total number of unique notes
    network_input, network_output = prepare_sequences(notes, n_vocab)  # Prepare sequences
    print(f"Prepared {len(network_input)} input sequences.")
else:
    print("Not enough notes to prepare sequences. Please check the dataset.")
```

```python
import pandas as pd
import numpy as np

def oversample(network_input, network_output, sequence_length=15):
    n_patterns = len(network_input)

    # Flatten network_input from (n_patterns, sequence_length, 1) to (n_patterns, sequence_length)
    network_input_flattened = network_input.reshape(n_patterns, sequence_length)

    # Create a DataFrame from the input and output
    new_df = pd.concat([pd.DataFrame(network_input_flattened), pd.DataFrame(network_output)], axis=1)

    # Rename columns for easy access
    new_df.columns = [x for x in range(sequence_length)] + ['Notes']
    print('Before oversampling:')
    print(new_df['Notes'].value_counts())  # Check class distribution

    # Oversampling logic
    max_class_size = 700  # Desired size for each class
    class_subsets = [new_df.query(f'Notes == {i}') for i in range(len(new_df["Notes"].unique()))]

    for i in range(len(new_df['Notes'].unique())):
        try:
            class_subsets[i] = class_subsets[i].sample(max_class_size, random_state=42, replace=True)
        except ValueError:
            print(f"Class {i} does not have enough samples for oversampling. Using all available samples.")

    oversampled_df = pd.concat(class_subsets, axis=0).sample(frac=1.0, random_state=42).reset_index(drop=True)
    print('After oversampling:')
    print(oversampled_df['Notes'].value_counts())  # Check class distribution again

    # Sample n_patterns from the oversampled DataFrame
    sampled_df = oversampled_df.sample(n_patterns, replace=True)

    # Convert the input columns back to a 3D array
    network_in = sampled_df[[x for x in range(sequence_length)]]
    network_in = np.array(network_in)
    network_in = np.reshape(network_in, (n_patterns, sequence_length, 1))
    network_in = network_in / len(set(new_df['Notes']))  # Normalization

    # Convert the target column into a one-hot encoded matrix
    network_out = pd.get_dummies(sampled_df['Notes'])

    return network_in, network_out

# Example initialization
n_patterns = 1000  # Number of sequences
sequence_length = 15  # Length of each sequence
num_classes = 10  # Number of unique classes in `networkOutput`

# Random input data: (n_patterns, sequence_length, 1)
networkInput = np.random.randint(0, 100, size=(n_patterns, sequence_length, 1))

# Random output labels: (n_patterns,)
networkOutput = np.random.randint(0, num_classes, size=(n_patterns,))

# Call the oversample function
networkInputShaped, networkOutputShaped = oversample(networkInput, networkOutput, sequence_length=sequence_length)

print("Shape of oversampled network input:", networkInputShaped.shape)
print("Shape of oversampled network output:", networkOutputShaped.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Activation

# Define parameters
sequence_length = 32  # Sequence length
num_features = 1  # Number of features per time step (normalized input)
n_vocab = len(set(notes))  # Total number of unique notes

# Build the model
model = Sequential()
model.add(LSTM(512, input_shape=(sequence_length, num_features), return_sequences=True))
model.add(Dropout(0.3))
model.add(LSTM(512, return_sequences=False))
model.add(Dropout(0.3))
model.add(Dense(256, activation='relu'))
model.add(Dropout(0.3))
model.add(Dense(n_vocab))  # Output layer with n_vocab neurons
model.add(Activation('softmax'))  # Softmax for probabilities

# Compile the model
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
print(model.summary())

print(f"Model output size: {model.output_shape[-1]}, Expected n_vocab: {n_vocab}")
n_vocab = len(set(notes))  # Number of unique notes
print(f"Number of unique notes: {n_vocab}")
print(f"Sample notes: {notes[:10]}")
prediction_output = generate_notes(model, network_input, pitchnames, n_vocab, num_notes=500, temperature=1.2)
print("Generated Notes Sequence:", prediction_output[:20])
```

```python
import pretty_midi
import os

def load_notes_from_midi(directory):
    all_notes = []
    for filename in os.listdir(directory):
        if filename.endswith('.mid'):
            midi_path = os.path.join(directory, filename)
            midi_data = pretty_midi.PrettyMIDI(midi_path)
            for instrument in midi_data.instruments:
                for note in instrument.notes:
                    all_notes.append(note.pitch)  # Use note.name if you prefer note names
    return all_notes

# Load all notes from the /Songs directory
directory = '/content/drive/MyDrive/Music_composer/Songs'
allNotes = load_notes_from_midi(directory)
```

```python
import numpy as np
#Run this
def generate_notes(model, network_input, pitchnames, n_vocab, num_notes=500, temperature=1.0):
    """
    Generate a sequence of notes from the neural network with randomness.

    Parameters:
        model: Trained neural network model.
        network_input: Input data for the model.
        pitchnames: List of unique notes in the dataset.
        n_vocab: Total number of unique notes (vocabulary size).
        num_notes: Number of notes to generate.
        temperature: Controls randomness. Higher values = more randomness.

    Returns:
        prediction_output: List of generated notes/chords.
    """
    # Validate inputs
    if n_vocab == 0:
        raise ValueError("n_vocab is 0. Ensure 'notes' contains valid elements.")

    # Pick a random sequence from the input as a starting point for the prediction
    start = np.random.randint(0, len(network_input) - 1)
    print(f'Start index: {start}')
    int_to_note = {number: note for number, note in enumerate(pitchnames)}

    # Initialize the pattern from a random sequence in the input
    pattern = network_input[start].flatten().tolist()  # Flatten and convert to a list
    prediction_output = []

    # Generate the required number of notes
    for note_index in range(num_notes):
        # Reshape the pattern into the required input shape
        prediction_input = np.reshape(pattern, (1, len(pattern), 1))
        # Normalize the input
        prediction_input = prediction_input / float(n_vocab)

        # Predict the probabilities for the next note
        prediction = model.predict(prediction_input, verbose=0).flatten()

        # Ensure the prediction size matches n_vocab
        if len(prediction) != n_vocab:
            raise ValueError(
                f"Prediction size ({len(prediction)}) does not match n_vocab ({n_vocab}). "
                "Ensure the model's output layer matches the vocabulary size."
            )

        # Apply temperature scaling for randomness
        prediction = np.log(prediction + 1e-9) / temperature  # Prevent log(0) with small constant
        prediction = np.exp(prediction)  # Exponentiate to get scaled probabilities
        prediction /= np.sum(prediction)  # Normalize to sum to 1

        # Randomly sample a note index based on the probabilities
        index = np.random.choice(range(n_vocab), p=prediction)

        # Map the index to a note or chord
        if index in int_to_note:
            result = int_to_note[index]
            prediction_output.append(result)  # Append the result to the output sequence
        else:
            print(f"Warning: index {index} not in int_to_note.")
            result = None  # Placeholder for missing notes
            prediction_output.append("N/A")  # Mark invalid results in the sequence

        # Update the pattern for the next prediction
        pattern.append(index)  # Append the index (integer)
        pattern = pattern[1:]  # Drop the first element to maintain sequence length

    return prediction_output

# Example usage:
if len(notes) > 0:
    n_vocab = len(set(notes))
    pitchnames = sorted(set(notes))
    prediction_output = generate_notes(model, network_input, pitchnames, n_vocab, num_notes=500, temperature=1.2)
    print("Generated Notes Sequence (first 20):", prediction_output[:20])
else:
    print("Notes list is empty. Cannot generate notes.")
```

```python
import os
from music21 import note, chord, stream, instrument

def create_midi(prediction_output, output_file='/content/drive/MyDrive/Music_composer/Output/generated_music.mid'):
    offset = 0
    output_notes = []

    # Ensure the output directory exists
    output_directory = os.path.dirname(output_file)
    if not os.path.exists(output_directory):
        os.makedirs(output_directory)  # Create the directory if it doesn't exist

    # Create note and chord objects based on the values generated by the model
    for pattern in prediction_output:
        pattern_str = str(pattern)  # Convert pattern to string

        # pattern is a chord
        if ('.' in pattern_str) or pattern_str.isdigit():
            try:
                notes_in_chord = pattern_str.split('.')
                notes = [note.Note(int(n)) for n in notes_in_chord]
                for new_note in notes:
                    new_note.storedInstrument = instrument.Piano()
                new_chord = chord.Chord(notes)
                new_chord.offset = offset
                output_notes.append(new_chord)
            except Exception as e:
                print(f"Skipping invalid chord pattern: {pattern} due to {e}")
        else:
            try:
                new_note = note.Note(pattern_str)
                new_note.offset = offset
                new_note.storedInstrument = instrument.Piano()
                output_notes.append(new_note)
            except Exception as e:
                print(f"Skipping invalid note: {pattern} due to {e}")

        offset += 0.5

    midi_stream = stream.Stream(output_notes)
    midi_stream.write('midi', fp=output_file)
    print(f"MIDI file created: {output_file}")

# Generate MIDI:
if len(prediction_output) > 0:
    create_midi(prediction_output, output_file='/content/drive/MyDrive/Music_composer/Output/generated_music.mid')
else:
    print("Prediction output is empty. Cannot create MIDI file.")
```

```python
from music21 import note, chord, stream, instrument
import os

def create_midi(prediction_output, output_file='/content/drive/MyDrive/Music_composer/Output/generated_music.mid'):
    offset = 0
    output_notes = []

    # Ensure the output directory exists
    output_directory = os.path.dirname(output_file)
    if not os.path.exists(output_directory):
        os.makedirs(output_directory)

    # Create note and chord objects based on the values generated by the model
    for pattern in prediction_output:
        pattern_str = str(pattern)  # Convert pattern to string

        # pattern is a chord
        if ('.' in pattern_str) or pattern_str.isdigit():
            notes_in_chord = pattern_str.split('.')
            notes = []
            for current_note in notes_in_chord:
                new_note = note.Note(int(current_note))
                new_note.storedInstrument = instrument.Piano()
                notes.append(new_note)
            new_chord = chord.Chord(notes)
            new_chord.offset = offset
            output_notes.append(new_chord)
        # pattern is a single note
        else:
            try:
                # Try to create a note by converting to integer (for MIDI values)
                new_note = note.Note(int(pattern))
            except ValueError:
                # If pattern is not an integer, assume it is a note name
                new_note = note.Note(pattern)

            new_note.offset = offset
            new_note.storedInstrument = instrument.Piano()
            output_notes.append(new_note)

        # Increase offset each iteration so that notes do not stack
        offset += 0.5

    midi_stream = stream.Stream(output_notes)
    midi_stream.write('midi', fp=output_file)
    print(f"MIDI file created: {output_file}")

# Define the path where you want to save the MIDI file
output_file_path = '/content/drive/MyDrive/Music_composer/Output/generated_music.mid'

# Call the function to create and save the MIDI file
create_midi(prediction_output, output_file_path)
```
